```python
from tensorflow.keras import Sequential
```

```python
from tensorflow.keras.applications import ResNet50
```

```python
model = ResNet50()
```

```python
model.summary()
```

```python
from tensorflow.keras.utils import plot_model
```

```python
plot_model(model,
           show_shapes =  True,
       show_layer_names = True,
          show_dtype = True,show_layer_activations = True, expand_nested  = True )
```

```python
model = ResNet50(weights='imagenet',include_top=False, input_shape = (224,224,3))
```

```python
model.summary()
```

```python
plot_model(model,
           show_shapes =  True,
       show_layer_names = True,
          show_dtype = True,show_layer_activations = True, expand_nested  = True )
```

```python
from tensorflow.keras import Model #Functional implementation of keras
```

```python
output = model.output
```

```python
from tensorflow.keras.layers import Conv2D, MaxPool2D, Flatten, Dense
```

```python
output = MaxPool2D(pool_size = (2,2))(output)
```

```python
output = Flatten()(output)
```

```python
output = Dense(units =500,
                 kernel_initializer = "glorot_uniform",
                 bias_initializer = "glorot_uniform",
                 activation = "relu",
                 name = "fc_layer_1")(output)
```

```python
#output layer \

output = Dense(units =1,
                 kernel_initializer = "glorot_uniform",
                 bias_initializer = "glorot_uniform",
                 activation = "sigmoid",
                 name = "output_layer")(output)
```

```python
model.input
```

```python

#Functional implementation of Keras
custom_model = Model(inputs=model.input, outputs = output)
```

```python
custom_model
```

```python
plot_model(custom_model,
           show_shapes =  True,
       show_layer_names = True,
          show_dtype = True,show_layer_activations = True, expand_nested  = True )
```

```python
from tensorflow.keras.optimizers import Adam
adam = Adam(learning_rate=0.001)
```

```python
custom_model.compile(loss = "binary_crossentropy", optimizer = adam, metrics = ["accuracy"])
```
